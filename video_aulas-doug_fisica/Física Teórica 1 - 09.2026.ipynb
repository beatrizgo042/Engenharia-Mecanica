## 🎲 Sistema Internacional de unidades (SI)

Sistema MKS (Metro, kg, segundo)

#### Tempo: segundo (s)
    1min = 1. 60s
    1h = 1. 60min = 1. 60. 60s = 3.600s`

#### Comprimento: metro (m)
    2km = 2.000m
    1cm = 1/100m = 0,1m
    1mm = 1/1.000m = 0,001m

#### Massa: quilograma (kg)
    1g = 1/1.000kg = 0,001kg
    1mg = 1/1.000g = 1/1.000.000kg = 0,000001kg

#### Corrrente Elétrica: Ampere (A)
    1mA = 1/1.000A = 0,001A
    1µA = 1/1.000mA = 1/1.000.000A = 0,000001A

#### Temperatura Kelvin (K)
    1ºC = 1ºK
    температур Kelvin = temperatura ºC + 273,15
    температур ºC = temperatura Kelvin - 273,15

#### Quantidade de substância: mol (mol)
    1mol = 6.022 * 10^23 átomos
    1mmol = 1/1.000mol = 0,001mol
    1µmol = 1/1.000mmol = 1/1.000.000mol = 0,000001mol

#### Intensidade luminosa: candela (cd)
    1mcd = 1/1.000cd = 0,001cd
    1µcd = 1/1.000mcd = 1/1.000.000cd = 0,000001cd

## 💻 Conversão de Unidades

#### Velocidade

> Velocidade: km/h -> m/s <br>
> Velocidade = V = Δx/Δt
        Δx = 4m
        Δt = 2s
        V = 4/2 = 2m/s

**Ex1:** 
```
    1km/h = 1.000m/3.600s 
    1km/h = (1.000/3.600) * (km/s) = 
    1km/h 10/36 * (km/s) 
    1km/h = 0,27777777777777777777777777777778m/s
```

**Ex2:**
```
    m/s -> km/h
    1m/s = 1 * 0,001km/ ((1/3600)*h) = 0,001km * 3600/h = 3.6km/h
    
    m/h * 3,6
    Km/h / 3,6
```

## 👩🏻‍💻 Operações com Expoentes

Multiplicação de Expoentes
```
10^4 * 10^1 = 10^(4+1) = 10^5
10^2 * 10^7 = 10^(2+7) = 10^9
```

Divisão de Expoentes
```
10^4 / 10^1 = 10^(4-1) = 10^3
10^2 / 10^7 = 10^(2-7) = 10^-5
```

Potenciação de Expoentes
```
10^4 * 10^4 = 10^(4*4) = 10^16
10^2 * 10^7 = 10^(2*7) = 10^14
```

Raiz de Expoentes
```
√10^4 = 10^(4/2) = 10^2
√10^2 = 10^(2/2) = 10^1
```

Exponenciação de Expoentes
```
10^4 * 10^4 = 10^(4*4) = 10^16
10^2 * 10^7 = 10^(2*7) = 10^14
```
